# Multiple Linear Regression – Quiz Questions

## Setup
Import the required packages and generate the dataset.

```python
import numpy as np
from pandas import DataFrame
import plotly.express as px
from patsy import dmatrices
import statsmodels.api as sm

# Seed the pseudo-random number generator
np.random.seed(12)

# Sample size
n = 100

# Create data
wcc = np.round(np.random.normal(15, 5, n), 0)

crp = (wcc * 2) + np.round(
    np.random.normal(0, 10, n), 0
)

lipase = (
    wcc +
    crp +
    np.round(np.random.normal(2, 10, n), 0)
)

# Create dataframe
df = DataFrame(
    {
        'WCC': wcc,
        'CRP': crp,
        'Lipase': lipase
    }
)

# Display first rows
df.head()
```

## Visualize the Data
Create a scatter plot matrix of the variables.

```python
px.scatter_matrix(
    df,
    title='Scatter Plot Matrix'
)
```

## Question 1
Create a multiple linear regression model with WCC and CRP as predictors of Lipase.

```python
# Create target vector and design matrix
y, X = dmatrices(
    'Lipase ~ WCC + CRP',
    data=df
)

# Fit regression model
model = sm.OLS(y, X).fit()

# View summary
model.summary()
```

## Question 2
Comment on the F statistic.

### Answer

The F statistic tests whether the regression model as a whole is statistically significant.

- A large F statistic with a very small p-value (typically p < 0.05)
  indicates that the predictors together significantly explain variation in Lipase.

- In this model, the F statistic is statistically significant,
  meaning that WCC and CRP together are useful predictors of Lipase.

## Question 3
Comment on the individual coefficients and their p values.

### Answer

The regression coefficients represent the expected change in Lipase
for a one-unit increase in each predictor while holding the other predictor constant.

- WCC coefficient:
  Indicates how much Lipase changes for each unit increase in WCC.

- CRP coefficient:
  Indicates how much Lipase changes for each unit increase in CRP.

The p-values test whether each coefficient is significantly different from zero.

- If p < 0.05:
  The predictor is statistically significant.

- If p > 0.05:
  The predictor is not statistically significant.

In this model, both WCC and CRP are expected to have significant positive relationships with Lipase because the data was generated using these variables.

## Question 4
Comment on the R² value.

### Answer

The R² value measures the proportion of variation in Lipase
explained by the predictors WCC and CRP.

- R² ranges from 0 to 1.
- A higher R² indicates better model fit.

For example:

- R² = 0.80 means that 80% of the variation in Lipase
  is explained by WCC and CRP.

In this model, the R² value is expected to be relatively high
because Lipase was generated directly from WCC and CRP.

predict Lipase values.

- Individual coefficients evaluate the contribution of each predictor.
- p-values determine statistical significance.
- R² measures how much variation in Lipase is explained by the model.
